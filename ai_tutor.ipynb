```python
#Import the neccesary libraries
import os
from google import genai
from dotenv import load_dotenv
from google.genai import types
import gradio as gr
```

```python
#Generate an gemini api key in Google AI Studio
#create a .env file and save the key as GEMINI_API_KEY=AIzaSyDxiQxxxxxxxxxxx....
#load the key as environment varible

load_dotenv()
gemini_api_Key=os.getenv("GEMINI_API_KEY")
print("Gemini API key loaded successfully.")
print(f"Key starts with : {gemini_api_Key[:10]}")
```

```python
#Configure the Gemini Client using the loaded key
gemini_client=genai.Client(api_key=gemini_api_Key)
print("Gemini client configured")
```

```python
# Define the mapping for explanation levels
explanation_levels = {
    1: "like I'm 5 years old",
    2: "like I'm 10 years old",
    3: "like a high school student",
    4: "like a college student",
    5: "like an expert in the field",
    6:"like an Einstien PHD-level mad scientist"
}
```

```python
def ai_tutor_with_level(user_question,level):
    exp_level=explanation_levels.get(level, "cleary and concisely")
    system_prompt=f"You're a helpful AI tutor. Explain the concpets {exp_level}"
    try:
        stream=gemini_client.models.generate_content_stream(
            model="gemini-3.5-flash",
            config=types.GenerateContentConfig(
                system_instruction=system_prompt,
                temperature=0.1),
            contents=user_question)
        full_response=""
        for chunk in stream:
            if chunk.text:
                full_response+=chunk.text
                yield full_response

    except Exception as e:
        print(f"An error occured during streaming {e}")
        yield f"Sorry I encountered an error {e}"
```

```python
ai_tutor_interface_slider= gr.Interface(fn=ai_tutor_with_level,
    inputs=[
        gr.Textbox(lines=2, label="Your Question" , placeholder="Ask AI tutor a question..."),
        gr.Slider(
            minimum=1,
            maximum=6,
            step=1,
            value=3,
            label="Explanation Level"),],
    outputs=gr.Markdown(label="AI Tutor's Explanation", container=True, height=300),
    title= "🎓 Advanced AI Tutor",
    description=" Ask a question and select the desired explanation level using slider.",
    flagging_mode="never")
print("Launching Advanced Gradio Interface with Slider...")
ai_tutor_interface_slider.launch()
```
